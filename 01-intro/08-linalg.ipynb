```python
import numpy as np
```

```python
def dot_product(u, v):
    assert u.shape[0] == v.shape[0]

    n = u.shape[0]

    result = 0.0
    for i in range(n):
        result = result + u[i] * v[i]

    return result
```

```python
u = np.array([1, 2, 3, 4])
v = np.array([5, 6, 7, 8])
```

```python
dot_product(u, v)
```

```python
u.dot(v)
```

```python
U = np.array([
    [2, 4, 5, 6],
    [1, 2, 1, 2],
    [3, 1, 2, 1]
])
```

```python
def matrix_vector_multiplication(U, v):
    assert U.shape[1] == v.shape[0]

    numb_rows = U.shape[0]

    result = np.zeros(numb_rows)
    for i in range(numb_rows):
        result[i] = dot_product(U[i], v)

    return result
```

```python
matrix_vector_multiplication(U, v)
```

```python
U.dot(v)
```

```python
U @ v
```

```python
def matrix_matrix_multiplication(U, V):
    assert U.shape[1] == V.shape[0]
    numb_rows = U.shape[0]
    numb_columns = V.shape[1]
    result = np.zeros((numb_rows, numb_columns))

    for i in range(numb_columns):
        result[:, i] = matrix_vector_multiplication(U, V[:, i])

    return result
```

```python
V = np.array([
    [1, 2, 3, 1],
    [4, 6, 6, 3],
    [3, 8, 9, 3],
    [10, 11, 12, 4]
])
```

```python
matrix_matrix_multiplication(U, V)
```

```python
J = U @ V
J
```

```python
U.dot(V)
```
